## Exercise 1: Traditional vs. Modern NLP

| Aspect | Traditional NLP | Modern NLP / LLM-based NLP |
|---|---|---|
| Feature engineering | Mostly manual. Humans design features such as word counts, TF-IDF scores, n-grams, sentiment lexicons, or grammatical rules. | Mostly automatic. Deep models learn useful features directly from data through embeddings and neural layers. |
| Word representations | Static representations. A word usually has the same representation regardless of context. | Contextual representations. The same word can have different representations depending on the sentence. |
| Model architectures | Shallow models such as Naive Bayes, logistic regression, SVMs, decision trees, or rule-based systems. | Deep neural architectures, especially Transformers with many layers and attention mechanisms. |
| Training methodology | Usually trained from scratch for one specific task using labeled data. | Usually pre-trained on massive text corpora, then fine-tuned or prompted for downstream tasks. |
| Examples | Bag-of-Words, TF-IDF, Naive Bayes, Hidden Markov Models, CRFs, SVM classifiers. | BERT, GPT, T5, RoBERTa, LLaMA-style decoder models, encoder-decoder translation models. |
| Advantages | Simple, fast, interpretable, works well with small datasets and limited compute. | More accurate, adaptable, scalable, better at context, supports many tasks with one model family. |
| Disadvantages | Requires heavy manual feature design, struggles with context, ambiguity, and transfer between tasks. | Expensive to train, less interpretable, may inherit bias, can hallucinate, requires careful deployment. |

The shift from traditional NLP to modern NLP greatly improved scalability because the same pre-trained model can be reused across many tasks instead of building a separate feature pipeline for each problem. It also improved efficiency at the application level: once a model is pre-trained, teams can fine-tune it with much less labeled data and adapt it faster to classification, summarization, question answering, search, or generation. However, this efficiency comes with a new cost: large models require more memory, compute, monitoring, and ethical safeguards.

## Exercise 2: LLM Architecture and Application Scenarios

| Architecture | Core architectural idea | Real-world application | Why it fits |
|---|---|---|---|
| BERT | Encoder-only Transformer. It reads the full context bidirectionally and is trained mainly with Masked Language Modeling. | Sentiment analysis or legal document classification. | Classification tasks benefit from understanding the whole sentence or document at once. BERT can use both left and right context before deciding the label. |
| GPT | Decoder-only Transformer. It reads previous tokens and predicts the next token using Causal Language Modeling. | Chatbots, creative writing, code generation, email drafting. | Generation tasks require producing text step by step. GPT-style models are naturally suited to continuing a sequence coherently. |
| T5 | Encoder-decoder Transformer. It converts every task into text-to-text format. | Translation, summarization, question answering, data-to-text generation. | The encoder understands the input text, while the decoder generates a transformed output. This is ideal when the output is a new text derived from the input. |

In short: BERT is strongest when the task is understanding, GPT is strongest when the task is open-ended generation, and T5 is strongest when the task requires transforming one piece of text into another.

## Exercise 3: Benefits and Ethical Considerations of Pre-training

**Five key benefits of pre-trained models**

1. **Improved generalization**: pre-trained models learn broad language patterns from large datasets, so they usually perform better on new examples than models trained only on a small task dataset.
2. **Reduced need for labeled data**: because the model already knows a lot about grammar, meaning, and common facts, fine-tuning can work with fewer labeled examples.
3. **Faster fine-tuning**: instead of learning language from zero, the model only needs to adapt its existing knowledge to the target task.
4. **Transfer learning**: knowledge learned from one large general training process can transfer to many downstream tasks such as classification, summarization, search, or question answering.
5. **Robustness**: exposure to diverse text makes the model more capable of handling different writing styles, domains, and vocabulary.

**Ethical concerns and mitigations**

| Concern | Risk | Mitigation strategy |
|---|---|---|
| Bias | The model may reproduce stereotypes or unfair associations from the training data. | Audit datasets, evaluate outputs across demographic groups, apply bias tests, and use human review for sensitive use cases. |
| Misinformation | The model may generate confident but false claims. | Use retrieval from trusted sources, add fact-checking, cite sources, and restrict use in high-stakes domains. |
| Privacy | Training data may contain personal or confidential information. | Filter sensitive data, use privacy-preserving training methods, and avoid exposing raw training examples. |
| Misuse | Models can be used to create spam, phishing, fake reviews, or harmful content. | Add usage policies, rate limits, abuse monitoring, safety classifiers, and access controls. |
| Environmental cost | Large-scale training consumes energy and hardware resources. | Reuse existing models, fine-tune efficiently, distill models, and choose smaller models when they are sufficient. |

## Exercise 4: Transformer Architecture Deep Dive

### Self-Attention and Multi-Head Attention

Self-attention allows every token in a sequence to compare itself with every other token. For each token, the model creates three vectors: a query, a key, and a value. The query asks what the token is looking for, the keys describe what each token contains, and the values carry the information that will be passed forward. The model calculates attention scores between queries and keys, normalizes them with softmax, and uses those weights to combine the value vectors. This gives each token a context-aware representation.

Multi-head attention repeats this process several times in parallel. Each head can specialize in different relationships: grammar, long-distance dependencies, named entities, pronouns, topic links, or semantic similarity. Compared with single-head attention, multi-head attention captures richer patterns because it does not force the model to represent every relationship through one attention map.

**Example sentence:** The doctor called the patient because she was worried.

One attention head may focus on the relation between **she** and **doctor** to resolve who was worried. Another head may focus on **called** and **patient** to understand the action. A third head may focus on **because** to identify the causal relation between the two clauses. Together, the heads build a more complete interpretation than a single attention pattern could.

### Pre-training Objectives

| Objective | Description | Best suited for |
|---|---|---|
| Masked Language Modeling | Some input tokens are hidden, and the model learns to predict the missing tokens using both left and right context. | Understanding tasks such as classification, entity recognition, semantic search, and reading comprehension. |
| Causal Language Modeling | The model predicts the next token using only previous tokens. | Generation tasks such as chat, story writing, code completion, and text continuation. |

MLM is more appropriate for a system that needs to understand a full document before assigning a label, such as detecting whether a review is positive or negative. CLM is more appropriate for a system that must generate the next part of a sequence, such as a chatbot response.

Early BERT models used Next Sentence Prediction to teach the model whether two sentences logically followed each other. This was meant to help tasks involving sentence pairs, such as question answering or natural language inference. Modern models often avoid NSP because later research showed that it did not always improve downstream performance, and that better training data, longer training, and improved masking strategies can be more useful.

### Transformer Model Selection

| Application | Best model type | Justification |
|---|---|---|
| Customer review sentiment analysis | Encoder-only | The task requires understanding the whole review and assigning a label. Encoder models are efficient and strong for classification. |
| Creative conversational chatbot | Decoder-only | The task requires generating natural responses token by token. Decoder models are designed for open-ended text generation. |
| English-to-Spanish technical translation | Encoder-decoder | The source text must be encoded, then a new target-language sequence must be generated. Encoder-decoder models are ideal for sequence-to-sequence transformation. |

### Positional Encoding

Transformers process tokens in parallel, so they do not naturally know the order of words. Positional encoding adds information about token position so the model can distinguish word order. Without positional encoding, sentences such as **The dog chased the cat** and **The cat chased the dog** could look too similar, even though their meanings are opposite.

## Exercise 5: BERT Variations - Choose Your Detective

### Scenario choices

| Scenario | Best BERT variation | Why |
|---|---|---|
| Real-time sentiment analysis on a mobile app with limited resources | DistilBERT | It is smaller and faster than BERT while keeping much of the performance, which makes it suitable for low-latency mobile inference. |
| Research on legal documents requiring high accuracy | RoBERTa | It improves BERT training by using more data, longer training, and better masking, often producing stronger accuracy. |

| Efficient pretraining and token replacement detection | ELECTRA | It trains by detecting replaced tokens rather than only predicting masked tokens, making pre-training more sample-efficient. |
| Efficient NLP in resource-constrained environments | ALBERT | It reduces parameters through factorized embeddings and parameter sharing, making it more memory-efficient. |

### Comparison table

| Model | Training data and method | Model size and efficiency | Key optimization | Ideal use cases |
|---|---|---|---|---|
| RoBERTa | Trained like BERT but with more data, dynamic masking, no NSP, larger batches, and longer training. | Usually larger and more compute-heavy than DistilBERT or ALBERT, but very accurate. | Stronger pre-training recipe. | High-accuracy classification, research, document understanding. |
| ALBERT | BERT-style training with parameter sharing and factorized embeddings. | Much fewer parameters than BERT, better memory efficiency. | Cross-layer parameter sharing and smaller embedding parameterization. | Efficient deployment where memory matters, classification, QA. |

| ELECTRA | Uses a generator-discriminator setup and replaced token detection. | More sample-efficient pre-training than MLM. | Learns from every token by detecting replacements. | Efficient pre-training, strong performance with less compute. |
| XLM-RoBERTa | Multilingual RoBERTa-style training on large multilingual corpora. | Larger model, but useful across many languages. | Cross-lingual representation learning. | Multilingual classification, global support, cross-lingual search. |

## Exercise 6: Softmax Temperature - The Randomness Regulator

### Temperature scenarios

| Temperature | Expected behavior |
|---|---|
| 0.2 | Very deterministic. The highest-probability tokens become even more dominant, so the output is focused, repetitive, and conservative. |
| 1.0 | Neutral setting. The model samples from the original probability distribution without sharpening or flattening it. |
| 1.5 | More random and creative. Lower-probability tokens become more likely, so the output can be diverse but also less coherent or less accurate. |

### Application design

For personalized bedtime stories for children, I would use a moderate temperature, for example around **0.8 to 1.1**. This keeps the story creative enough to feel imaginative, but not so random that it becomes confusing or inappropriate. For younger children, I would keep the temperature lower to preserve simple structure and safe vocabulary. For older children or fantasy stories, I could raise it slightly to increase surprise and originality.

For financial report summaries, I would use a low temperature, for example **0.1 to 0.4**. The goal is accuracy, consistency, and reliability, not creativity. I would also combine the model with source-grounded retrieval, numerical checks, and strict instructions to avoid inventing facts.

### Temperature and bias

Temperature can affect bias because it changes which tokens are likely to be selected. A low temperature may repeatedly select the model's most common learned associations, which can reinforce stereotypes if the highest-probability completions are biased. A high temperature may produce more varied outputs, but it can also increase the chance of toxic, unfair, or unusual associations appearing.

**Practical example:** if a model completes the sentence **The nurse said that...**, a very low temperature might repeatedly choose stereotypical continuations if the training data contains them. A higher temperature may diversify the completions, but it does not remove the underlying bias. The better solution is bias evaluation, safer data, careful prompting, moderation, and output review in sensitive contexts.

```python
# Small demonstration: how temperature changes a probability distribution
import numpy as np

logits = np.array([4.0, 2.0, 1.0, 0.5])
tokens = ["accurate", "creative", "unexpected", "risky"]

def softmax_with_temperature(logits, temperature):
    scaled = logits / temperature
    exp_values = np.exp(scaled - np.max(scaled))
    return exp_values / exp_values.sum()

for temperature in [0.2, 1.0, 1.5]:
    probs = softmax_with_temperature(logits, temperature)
    print(f"\nTemperature = {temperature}")
    for token, prob in zip(tokens, probs):
        print(f"{token:10s}: {prob:.3f}")
```

Modern NLP moved from hand-built feature pipelines to reusable pre-trained Transformer models. Encoder-only models are best for understanding, decoder-only models are best for generation, and encoder-decoder models are best for text transformation. Pre-training gives LLMs strong transfer learning abilities, but it also creates ethical responsibilities around bias, privacy, misinformation, and misuse. Finally, softmax temperature is a useful control for balancing coherence and creativity, but it must be combined with broader safety and quality controls.
