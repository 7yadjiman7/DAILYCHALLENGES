```python
import numpy as np
from scipy import stats 
```

```python
sample_data = np.array([15,15,15,15,15,15, 26, 25, 28])
t_statistic, p_value = stats.ttest_1samp(sample_data, popmean=15)

print(f'la moyenne de sample_data {np.mean(sample_data)}')
print(f'la taille {len(sample_data)}')

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
