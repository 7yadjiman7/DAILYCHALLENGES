```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

## Exercise 1 : Advanced Matrix Operations

```python
np.random.seed(42)
matrix = np.random.randint(1, 20, size=(5, 5)).astype(float)

print("Original 5x5 matrix:")
print(matrix)

# 1. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(matrix)

print("\nEigenvalues:")
print(eigenvalues.round(4))
print("\nEigenvectors (columns):")
print(eigenvectors.round(4))

# 2. Matrix normalization (min-max scaling to [0, 1])
mat_min = matrix.min()
mat_max = matrix.max()
normalized = (matrix - mat_min) / (mat_max - mat_min)

print("\nNormalized matrix (min-max [0, 1]):")
print(normalized.round(4))
```

## Exercise 2 : Statistical Distribution Analysis

```python
np.random.seed(0)
data = np.random.normal(loc=50, scale=10, size=1000)

skewness = stats.skew(data)
kurtosis = stats.kurtosis(data)  # excess kurtosis (normal = 0)

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(data, bins=40, color='steelblue', edgecolor='none', density=True, alpha=0.8, label='Data')
x = np.linspace(data.min(), data.max(), 300)
ax.plot(x, stats.norm.pdf(x, data.mean(), data.std()), color='crimson', linewidth=2, label='Normal fit')
ax.axvline(data.mean(),   color='orange', linewidth=1.5, linestyle='--', label=f'Mean={data.mean():.2f}')
ax.axvline(np.median(data), color='green', linewidth=1.5, linestyle=':', label=f'Median={np.median(data):.2f}')
ax.set_title('Normal Distribution — Histogram')
ax.set_xlabel('Value')
ax.set_ylabel('Density')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Skewness : {skewness:.4f}")
print(f"  → Near 0: the distribution is approximately symmetric.")
print(f"Kurtosis : {kurtosis:.4f}  (excess kurtosis; normal = 0)")
print(f"  → Near 0: tails are similar to a normal distribution (mesokurtic).")
```

## Exercise 3 : Time Series Forecasting

```python
np.random.seed(5)
monthly_sales_data = np.random.randint(100, 500, size=12)
months = np.arange(1, 13)

# Linear regression via np.polyfit
coeffs = np.polyfit(months, monthly_sales_data, deg=1)
trend  = np.polyval(coeffs, months)

# Forecast next 3 months
future_months  = np.arange(13, 16)
forecast_vals  = np.polyval(coeffs, future_months)

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(months, monthly_sales_data, marker='o', color='steelblue', linewidth=1.5, label='Actual Sales')
ax.plot(months, trend, color='orange', linewidth=1.5, linestyle='--', label='Trend line')
ax.plot(future_months, forecast_vals, marker='s', color='crimson',
        linewidth=1.5, linestyle='--', label='Forecast (months 13–15)')
ax.axvline(12.5, color='gray', linewidth=1, linestyle=':')
ax.set_title('Monthly Sales — Linear Regression Forecast')
ax.set_xlabel('Month')
ax.set_ylabel('Sales')
ax.set_xticks(np.arange(1, 16))
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Trend equation: Sales = {coeffs[0]:.2f} * month + {coeffs[1]:.2f}")
for m, v in zip(future_months, forecast_vals):
    print(f"  Forecast month {m}: {v:.1f} units")
```

## Exercise 4 : Pandas Data Aggregation

```python
data = {
    'Product': ['Product A', 'Product B', 'Product C', 'Product A', 'Product B'],
    'Sales': [200, 150, 300, 250, 180],
    'Month': ['January', 'February', 'March', 'April', 'May']
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Group by Product and aggregate
grouped = df.groupby('Product')['Sales'].agg(
    total_sales='sum',
    mean_sales='mean',
    max_sales='max',
    count='count'
).reset_index()

print("\nAggregation by Product:")
print(grouped)

print(f"\nBest-selling product : {grouped.loc[grouped['total_sales'].idxmax(), 'Product']}")
print(f"Total sales overall  : {df['Sales'].sum()}")
```

## Exercise 5 : Complex Data Visualization

```python
np.random.seed(3)
x  = np.linspace(0, 10, 50)
y_line    = np.sin(x) + np.random.normal(0, 0.1, size=50)
y_scatter = np.random.rand(50) * 10
categories = ['A', 'B', 'C', 'D', 'E']
bar_values = np.random.randint(20, 100, size=5)
hist_data  = np.random.normal(5, 1.5, size=500)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Multi-Plot Layout', fontsize=14, fontweight='bold')

# Line plot
axes[0, 0].plot(x, y_line, color='steelblue', linewidth=1.5)
axes[0, 0].set_title('Line Plot — sin(x) + noise')
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('y')
axes[0, 0].grid(True, alpha=0.3)

# Scatter plot
sc = axes[0, 1].scatter(x, y_scatter, c=y_scatter, cmap='plasma', s=30, alpha=0.8)
fig.colorbar(sc, ax=axes[0, 1], label='Value')
axes[0, 1].set_title('Scatter Plot')
axes[0, 1].set_xlabel('x')
axes[0, 1].set_ylabel('y')
axes[0, 1].grid(True, alpha=0.3)

# Bar chart
axes[1, 0].bar(categories, bar_values, color='coral', edgecolor='white')
axes[1, 0].set_title('Bar Chart')
axes[1, 0].set_xlabel('Category')
axes[1, 0].set_ylabel('Value')
for i, v in enumerate(bar_values):
    axes[1, 0].text(i, v + 1, str(v), ha='center', fontsize=9)
axes[1, 0].grid(axis='y', alpha=0.3)

# Histogram
axes[1, 1].hist(hist_data, bins=30, color='seagreen', edgecolor='none', density=True, alpha=0.8)
x_norm = np.linspace(hist_data.min(), hist_data.max(), 200)
axes[1, 1].plot(x_norm, stats.norm.pdf(x_norm, hist_data.mean(), hist_data.std()),
                color='crimson', linewidth=2, label='Normal fit')
axes[1, 1].set_title('Histogram with Normal Fit')
axes[1, 1].set_xlabel('Value')
axes[1, 1].set_ylabel('Density')
axes[1, 1].legend()
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
