```python
import numpy as np
```

## Exercise 1 : Maximum Value Replacement in Matrix

```python
matrix = np.random.rand(5, 5).round(2)
matrix[np.unravel_index(matrix.argmax(), matrix.shape)] = 0
matrix
```

## Exercise 2 : Finding Common Values in Arrays

```python
arr1 = np.random.rand(5).round(2)
arr2 = np.random.rand(5).round(2)
common = np.intersect1d(arr1, arr2)
print(f"Array 1: {arr1}")
print(f"Array 2: {arr2}")
print(f"Common values: {common}")
```

## Exercise 3 : Sorting Array in Ascending and Descending Order

```python
arr = np.random.rand(10).round(2)
ascending = np.sort(arr)
descending = np.sort(arr)[::-1]
print(f"Ascending:  {ascending}")
print(f"Descending: {descending}")
```

## Exercise 4 : Matrix Rank Determination

```python
matrix = np.random.rand(4, 4)
rank = np.linalg.matrix_rank(matrix)
print(f"Rank: {rank}")
```

## Exercise 5 : Creating Border Array

```python
arr = np.zeros((5, 5), dtype=int)
arr[[0, -1], :] = 1
arr[:, [0, -1]] = 1
arr
```
