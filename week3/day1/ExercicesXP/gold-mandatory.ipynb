```python
import numpy as np
```

## Exercise 1 : Minimum and Maximum of Random Array

```python
arr = np.random.rand(5, 5)
print(f"Min: {arr.min():.2f}, Max: {arr.max():.2f}")
```

## Exercise 2 : Matrix Normalization

```python
matrix = np.random.rand(3, 3)
normalized = (matrix - matrix.mean()) / matrix.std()
normalized
```

## Exercise 4 : Matrix Multiplication

```python
A = np.random.rand(5, 3)
B = np.random.rand(3, 2)
result = A @ B
result.round(2)
```

## Exercise 5 : Matrix Multiplication (custom)

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

result = A @ B
print("Matrix A (2x3):")
print(A)
print("\nMatrix B (3x2):")
print(B)
print("\nMatrix Multiplication Result (2x2):")
print(result)
```
