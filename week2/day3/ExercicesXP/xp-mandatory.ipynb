## Exercise 2: Descriptive Statistics

```python
from scipy import stats
import numpy as np

data = [12, 15, 13, 12, 18, 20, 22, 21]

mean = np.mean(data)
median = np.median(data)
variance = stats.tvar(data)
std_dev = stats.tstd(data)

print("Mean:", mean)
print("Median:", median)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

## Exercise 3: Understanding Distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 50
std = 10

x = np.linspace(mean - 4 * std, mean + 4 * std, 200)
y = norm.pdf(x, mean, std)

plt.plot(x, y)
plt.title("Normal Distribution (mean=50, std=10)")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

## Exercise 4: T-Test Application

```python
import numpy as np
from scipy import stats

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = stats.ttest_ind(data1, data2)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("The difference between the two groups is statistically significant (p < 0.05).")
else:
    print("The difference between the two groups is not statistically significant (p >= 0.05).")
```

## Exercise 5: Linear Regression Analysis

```python
from scipy import stats
import numpy as np

house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

print("Slope:", slope)
print("Intercept:", intercept)

# Predict price for a 90 square meter house
size_to_predict = 90
predicted_price = slope * size_to_predict + intercept
print(f"Predicted price for {size_to_predict} sqm: {predicted_price:.2f}")

# Interpretation
print(f"\nInterpretation: For each additional square meter, the house price increases by approximately {slope:.2f} currency units.")
```

## Exercise 6: Understanding ANOVA

```python
from scipy import stats

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print("F-value:", f_value)
print("P-value:", p_value)

if p_value < 0.05:
    print("The fertilizers have significantly different effects on plant growth (p < 0.05).")
else:
    print("No significant difference between fertilizers (p >= 0.05).")

print("\nIf the P-value were greater than 0.05, we would fail to reject the null hypothesis,"
      " meaning there is no statistically significant difference between the fertilizers' effects on plant growth.")
```

## Exercise 8: Correlation Coefficients (Optional)

```python
import pandas as pd
from scipy import stats

data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

pearson_corr, pearson_p = stats.pearsonr(data['age'], data['income'])
spearman_corr, spearman_p = stats.spearmanr(data['age'], data['income'])

print(f"Pearson correlation:  r = {pearson_corr:.4f}, p-value = {pearson_p:.4f}")
print(f"Spearman correlation: r = {spearman_corr:.4f}, p-value = {spearman_p:.4f}")
```
