# Statistical Distributions and Hypothesis Testing

## Exercise 1: Multivariate Normal Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, multivariate_normal

# --- Univariate normal distribution ---
univariate_data = norm.rvs(size=1000)

# --- Multivariate normal distribution ---
mean = [0, 0]
cov = [[1, 0], [0, 1]]  # diagonal covariance, independent variables
multivariate_data = multivariate_normal.rvs(mean=mean, cov=cov, size=1000)
```

```python
# --- Visualization ---
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Univariate histogram
axes[0].hist(univariate_data, bins=40, color='steelblue', edgecolor='white', density=True)
x = np.linspace(-4, 4, 300)
axes[0].plot(x, norm.pdf(x), color='crimson', linewidth=2, label='PDF')
axes[0].set_title("Univariate Normal Distribution")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Density")
axes[0].legend()

# Multivariate scatter plot
axes[1].scatter(multivariate_data[:, 0], multivariate_data[:, 1],
                alpha=0.3, s=12, color='mediumpurple')
axes[1].set_title("Multivariate Normal Distribution (2D)")
axes[1].set_xlabel("Variable 1")
axes[1].set_ylabel("Variable 2")
axes[1].axhline(0, color='grey', linewidth=0.8, linestyle='--')
axes[1].axvline(0, color='grey', linewidth=0.8, linestyle='--')

plt.tight_layout()
plt.show()
```

```python
# --- 2D density surface of the multivariate distribution ---
x_grid, y_grid = np.mgrid[-4:4:0.05, -4:4:0.05]
pos = np.dstack((x_grid, y_grid))
rv  = multivariate_normal(mean=mean, cov=cov)
z   = rv.pdf(pos)

fig, ax = plt.subplots(figsize=(7, 6))
cp = ax.contourf(x_grid, y_grid, z, levels=20, cmap='viridis')
plt.colorbar(cp, ax=ax, label='Probability Density')
ax.set_title("2D PDF of the Multivariate Normal Distribution")
ax.set_xlabel("Variable 1")
ax.set_ylabel("Variable 2")
plt.tight_layout()
plt.show()
```

```python
# --- Comparison with a correlated multivariate distribution ---
cov_corr = [[1, 0.8], [0.8, 1]]  # positive correlation between variables
multivariate_corr = multivariate_normal.rvs(mean=mean, cov=cov_corr, size=1000)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].scatter(multivariate_data[:, 0], multivariate_data[:, 1],
                alpha=0.3, s=12, color='mediumpurple')
axes[0].set_title("Multivariate Normal (no correlation)")
axes[0].set_xlabel("Variable 1")
axes[0].set_ylabel("Variable 2")

axes[1].scatter(multivariate_corr[:, 0], multivariate_corr[:, 1],
                alpha=0.3, s=12, color='darkorange')
axes[1].set_title("Multivariate Normal (correlation = 0.8)")
axes[1].set_xlabel("Variable 1")
axes[1].set_ylabel("Variable 2")

plt.tight_layout()
plt.show()
```

### Differences between univariate and multivariate normal distributions

**Univariate normal distribution**
- Describes a single random variable.
- Fully characterized by two parameters: a mean (location) and a standard deviation (spread).
- The probability density function is the familiar symmetric bell curve.
- Visualized as a 1D histogram or density curve.

**Multivariate normal distribution**
- Generalizes the normal distribution to two or more random variables simultaneously.
- Characterized by a mean vector (one mean per variable) and a covariance matrix (variances on the diagonal, covariances off-diagonal).
- The covariance matrix encodes both the spread of each variable and the linear relationships between them.
- Visualized as a 2D scatter plot, contour plot, or 3D surface.

**Effect of correlation**
- When the covariance matrix is diagonal (off-diagonal terms = 0), the variables are independent and the scatter plot is roughly circular.
- When the off-diagonal terms are non-zero, the variables are correlated and the scatter plot is elongated along a diagonal axis, as shown in the comparison above.
- A positive covariance means the variables tend to increase together; a negative covariance means they move in opposite directions.

## Exercise 2: Advanced Probability Distributions

### Chosen distribution: Poisson Distribution
### Real-world scenario: Number of customer arrivals at a store

**Description**

A supermarket wants to plan its staffing levels throughout the day. Based on historical data, an average of 8 customers arrive every 10 minutes during peak hours. The Poisson distribution can model the probability of exactly k customers arriving in any given 10-minute window.

**Assumptions**
- Events (arrivals) occur independently of one another: the arrival of one customer does not affect the arrival of the next.
- The average rate of arrivals (lambda, λ = 8 per 10 min) remains constant throughout the time interval being considered.
- Two arrivals cannot happen at the exact same instant (events are singular in time).
- The distribution applies to a fixed, well-defined observation window (here, 10 minutes).

**Implications**
- The Poisson distribution gives the probability of observing exactly 0, 1, 2, … customers in the window.
- With λ = 8, the most likely counts are around 7–9 customers per interval.
- The store can calculate the probability of unusually high traffic (e.g., more than 15 customers in 10 minutes) to decide when to open additional checkout lanes.
- Because the variance of a Poisson distribution equals its mean (both equal λ), periods with higher average traffic also have proportionally higher variability in actual arrivals.
- The model breaks down if arrivals cluster (e.g., a tour bus arrives at once) or if the rate changes significantly within the interval, violating the independence and constant-rate assumptions.

## Exercise 3: Statistical Hypothesis Testing — ANOVA

```python
import numpy as np
import pandas as pd
from scipy.stats import f_oneway

# Seed for reproducibility
np.random.seed(0)

# Sales data across different regions
region1 = np.random.normal(20000, 3000, 30)
region2 = np.random.normal(22000, 3500, 30)
region3 = np.random.normal(25000, 5000, 30)

sales_data = pd.DataFrame({
    'Region 1': region1,
    'Region 2': region2,
    'Region 3': region3
})

print("Descriptive statistics per region:")
print(sales_data.describe().round(2))
```

```python
# --- Perform one-way ANOVA ---
f_stat, p_value = f_oneway(region1, region2, region3)

print(f"F-statistic : {f_stat:.4f}")
print(f"P-value     : {p_value:.4e}")

if p_value < 0.05:
    print("\nConclusion: At the 5% significance level, there is a statistically significant"
          " difference in average sales across the three regions (p < 0.05).")
    print("We reject the null hypothesis that all group means are equal.")
else:
    print("\nConclusion: There is no statistically significant difference in average sales"
          " across the three regions (p >= 0.05).")
    print("We fail to reject the null hypothesis.")
```

```python
# --- Visualization ---
import matplotlib.pyplot as plt
import seaborn as sns

sales_long = sales_data.melt(var_name='Region', value_name='Sales')

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Boxplot
sns.boxplot(data=sales_long, x='Region', y='Sales', palette='Set2', ax=axes[0])
axes[0].set_title("Sales Distribution by Region")
axes[0].set_ylabel("Sales")

# Means with error bars (95% CI)
means  = sales_data.mean()
errors = sales_data.sem() * 1.96
axes[1].bar(means.index, means.values, yerr=errors.values,
            color=['#66c2a5', '#fc8d62', '#8da0cb'], edgecolor='white', capsize=5)
axes[1].set_title("Mean Sales per Region (95% CI)")
axes[1].set_ylabel("Mean Sales")

plt.tight_layout()
plt.show()
```

### Interpretation of the ANOVA test

**What ANOVA tests**

A one-way ANOVA (Analysis of Variance) tests whether the means of three or more independent groups are equal. The null hypothesis H0 states that all group means are identical; the alternative hypothesis H1 states that at least one group mean differs.

**How it works**

ANOVA partitions the total variance in the data into two components:
- *Between-group variance*: how much the group means differ from the overall mean.
- *Within-group variance*: how much individual observations scatter around their own group mean.

The F-statistic is the ratio of between-group variance to within-group variance. A large F-statistic means the groups differ more than would be expected by chance alone.

**Results in context**

The three regions were generated with different true means (20,000 / 22,000 / 25,000). The ANOVA correctly detects this difference: the p-value is well below 0.05, leading us to reject the null hypothesis. This tells us that at least one region's average sales is significantly different from the others, which is consistent with the data-generating process.

**Limitation**

ANOVA only tells us that a difference exists somewhere among the groups; it does not identify which specific pairs differ. A post-hoc test (e.g., Tukey's HSD) would be needed to determine exactly which regions differ from each other.

## Exercise 4: Linear Regression Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Seed for reproducibility
np.random.seed(0)

# Sample data
X = np.random.rand(100) * 50          # Independent variable: hours studied
Y = 2.5 * X + np.random.randn(100) * 10  # Dependent variable: test scores

linear_regression_data = pd.DataFrame({
    'Hours Studied': X,
    'Test Score': Y
})

print("Dataset overview:")
print(linear_regression_data.describe().round(2))
```

```python
# --- Perform linear regression ---
slope, intercept, r_value, p_value, std_err = linregress(
    linear_regression_data['Hours Studied'],
    linear_regression_data['Test Score']
)

r_squared = r_value ** 2

print(f"Slope (coefficient) : {slope:.4f}")
print(f"Intercept           : {intercept:.4f}")
print(f"R (correlation)     : {r_value:.4f}")
print(f"R-squared           : {r_squared:.4f}")
print(f"P-value             : {p_value:.4e}")
print(f"Standard error      : {std_err:.4f}")
```

```python
# --- Predicted values and residuals ---
linear_regression_data['Predicted Score'] = (
    slope * linear_regression_data['Hours Studied'] + intercept
)
linear_regression_data['Residual'] = (
    linear_regression_data['Test Score'] - linear_regression_data['Predicted Score']
)

print("\nSample predictions:")
print(linear_regression_data.head(8).round(2))
```

```python
# --- Visualization ---
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Scatter plot with regression line
x_line = np.linspace(X.min(), X.max(), 200)
y_line = slope * x_line + intercept

axes[0].scatter(linear_regression_data['Hours Studied'],
                linear_regression_data['Test Score'],
                alpha=0.6, s=30, color='steelblue', label='Observations')
axes[0].plot(x_line, y_line, color='crimson', linewidth=2,
             label=f'y = {slope:.2f}x + {intercept:.2f}')
axes[0].set_title(f"Linear Regression (R² = {r_squared:.3f})")
axes[0].set_xlabel("Hours Studied")
axes[0].set_ylabel("Test Score")
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.4)

# Residual plot
axes[1].scatter(linear_regression_data['Predicted Score'],
                linear_regression_data['Residual'],
                alpha=0.6, s=30, color='mediumpurple')
axes[1].axhline(0, color='crimson', linewidth=1.5, linestyle='--')
axes[1].set_title("Residuals vs Fitted Values")
axes[1].set_xlabel("Predicted Score")
axes[1].set_ylabel("Residual")
axes[1].grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

### Interpretation of the linear regression

**Slope**

The slope is approximately 2.5, which means that for each additional hour studied, the predicted test score increases by about 2.5 points. This matches the true relationship used to generate the data (Y = 2.5X + noise), confirming that the regression successfully recovered the underlying signal.

**Intercept**

The intercept represents the predicted test score when a student studies 0 hours. Its value may be close to 0 or slightly positive/negative due to the random noise added to the data. In practice, its interpretation depends on whether a student with zero study hours is a meaningful case in the context.

**R-squared**

The R-squared value measures the proportion of variance in test scores explained by study hours. A value close to 1 indicates a strong linear relationship; a lower value reflects more unexplained variability (due to the noise term in our data). Here, with a standard deviation of 10 for the noise and a slope of 2.5 over a range of 0–50 hours, we expect a moderately high R-squared.

**P-value**

The p-value tests whether the slope is significantly different from zero. A p-value well below 0.05 confirms that the relationship between study hours and test scores is statistically significant and not due to chance.

**Residual plot**

The residuals (differences between observed and predicted values) should be randomly scattered around zero with no clear pattern. A pattern in the residual plot would indicate that a linear model is not the best fit. Here the residuals appear randomly distributed, validating the linearity assumption.
