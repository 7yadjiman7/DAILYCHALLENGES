## Exercise 1 : Introduction to Data Analysis

### What is Data Analysis?

Data analysis is the process of inspecting, cleaning, transforming, and modeling data with the goal of discovering useful information, drawing conclusions, and supporting decision-making. It involves applying statistical and logical techniques to describe, condense, recap, and evaluate data.

### Why is Data Analysis Important?

In the modern world, data is generated at an unprecedented rate. Data analysis allows organizations and individuals to:
- **Make informed decisions** based on evidence rather than intuition.
- **Identify trends and patterns** that would otherwise remain hidden in raw data.
- **Reduce costs and improve efficiency** by optimizing processes.
- **Predict future outcomes** using historical data.
- **Gain a competitive advantage** in business by understanding customer behavior and market dynamics.

### Three Areas Where Data Analysis is Applied Today

**1. Healthcare**  
Data analysis is used to predict disease outbreaks, personalize patient treatment plans, and improve hospital operations. Analyzing patient records and clinical data helps doctors diagnose conditions earlier and recommend more effective treatments.

**2. Finance and Banking**  
Banks and financial institutions use data analysis for fraud detection, credit risk assessment, and investment strategy. Algorithms analyze thousands of transactions in real time to flag suspicious activity and protect customers.

**3. Marketing and E-commerce**  
Companies like Amazon and Netflix analyze user behavior, purchase history, and browsing patterns to provide personalized recommendations, optimize pricing, and target advertising campaigns more effectively.

**1. `Year` + `Avg hrs per day sleeping` → Trend Analysis**

**2. `Sex` + `Avg hrs per day sleeping` → Group Comparison**  
Comparer la durée moyenne de sommeil entre hommes ("Men"), femmes ("Women") et la moyenne générale ("Both") permet d'identifier des différences comportementales significatives entre les sexes.

**3. `Age Group` + `Avg hrs per day sleeping` → Group Comparison**  
Les tranches d'âge (ex: "15 to 24 years", "65 years and over") permettent d'analyser comment le temps de sommeil évolue avec l'âge — une question clé en santé publique.

**4. `Type of Days` + `Avg hrs per day sleeping` → Group Comparison**  
Comparer le sommeil les jours de semaine ("Nonholiday weekdays") versus le week-end ("Weekend days and holidays") révèle le déficit de sommeil hebdomadaire, un phénomène bien documenté.


## Exercise 6 : Identifying Data Types — Structured vs Unstructured

| # | Source de données | Type | Justification |
|---|---|---|---|
| 1 | Rapports financiers dans un fichier Excel | **Structured** | Données organisées en lignes et colonnes avec un schéma défini |
| 2 | Photographies sur une plateforme sociale | **Unstructured** | Images sans schéma interne exploitable directement |
| 3 | Collection d'articles de presse | **Unstructured** | Texte libre sans format tabulaire standardisé |
| 4 | Données d'inventaire dans une base relationnelle | **Structured** | Organisées en tables avec clés et types de colonnes définis |
| 5 | Entretiens enregistrés (audio) | **Unstructured** | Audio brut sans organisation formelle exploitable directement |

```python
## Exercise 9 : Export a DataFrame to Excel and JSON

# Install openpyxl if not already present (required for Excel export)
import subprocess, sys
subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'openpyxl', '-q'])

import pandas as pd

# Create a simple DataFrame
df = pd.DataFrame({
    'Name':  ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age':   [25, 30, 35, 28],
    'City':  ['Paris', 'Lyon', 'Nice', 'Bordeaux'],
    'Score': [88.5, 92.0, 79.3, 95.1]
})

print("=== DataFrame créé ===")
print(df)
print()

# Export to Excel
df.to_excel('output.xlsx', index=False)
print("✅ Exporté en Excel : output.xlsx")

# Export to JSON
df.to_json('output.json', orient='records', indent=2)
print("✅ Exporté en JSON : output.json")
```
